```python
from fastapi import FastAPI
from pydantic import BaseModel
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier
import joblib
```

```python
# Load the Iris dataset
iris = load_iris()

# Features
X = iris.data

# Target
y = iris.target

print("Dataset loaded successfully!")
print("Number of samples:", len(X))
```

```python
# Create the model
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
model.fit(X, y)

print("Model trained successfully!")
```

```python
# Save the trained model
joblib.dump(model, "model.pkl")

print("Model saved as model.pkl")
```

```python
# Load the saved model
loaded_model = joblib.load("model.pkl")

print("Model loaded successfully!")
```

```python
# Create FastAPI app
app = FastAPI(
    title="Iris Prediction API",
    version="1.0"
)

print("FastAPI app created!")
```

```python
class IrisInput(BaseModel):
    sepal_length: float
    sepal_width: float
    petal_length: float
    petal_width: float

print("Input schema created!")
```

```python
@app.get("/")
def home():
    return {
        "message": "FastAPI Model Serving Endpoint is Running!"
    }

print("Home endpoint created!")
```

```python
@app.post("/predict")
def predict(data: IrisInput):

    features = [[
        data.sepal_length,
        data.sepal_width,
        data.petal_length,
        data.petal_width
    ]]

    prediction = loaded_model.predict(features)[0]

    flower_name = iris.target_names[prediction]

    return {
        "prediction": flower_name
    }

print("Prediction endpoint created!")
```

```python
test_inputs = [
    [5.1, 3.5, 1.4, 0.2],
    [6.0, 2.9, 4.5, 1.5],
    [6.5, 3.0, 5.8, 2.2]
]

for i, sample in enumerate(test_inputs, start=1):
    prediction = loaded_model.predict([sample])[0]
    flower = iris.target_names[prediction]

    print(f"Test {i}")
    print("Input:", sample)
    print("Prediction:", flower)
    print("-" * 30)
```
