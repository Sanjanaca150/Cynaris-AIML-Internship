```python
import pandas as pd
import numpy as np

data = {
    "Name":["Asha","Rahul","Priya",None],
    "Age":[21,22,None,20],
    "Marks":[85,90,88,None]
}

df = pd.DataFrame(data)
df

df.head()
df.info()

df.describe()

df.isnull()

df.isnull().sum()

df.fillna(0)

clean_df = df.fillna({
    "Name":"Unknown",
    "Age":0,
    "Marks":0
})

clean_df

clean_df.isnull().sum()

clean_df.to_csv("cleaned_data.csv", index=False)
```
