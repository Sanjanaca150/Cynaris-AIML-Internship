```python
import pandas as pd

df = pd.read_excel("Indian Population Dataset.xlsx")

print(df.shape)

print(df.dtypes)

print(df.head(10))

filtered_df = df[df["Population"]>70000000]

print(filtered_df)

grouped = df.groupby("Region")["Population"].sum()

print(grouped)

capital_df = pd.DataFrame({
    "State": ["Karnataka", "Tamil Nadu", "Kerala", "Maharashtra", "Delhi"],
    "Capital": ["Bengaluru", "Chennai", "Thiruvananthapuram", "Mumbai", "New Delhi"]
})

merged_df = pd.merge(df, capital_df, on="State", how="left")

print(merged_df)

pivot = pd.pivot_table(
    df,
    values="Population",
    index="Region",
    aggfunc="sum"
)

print(pivot)

df.to_csv("cleaned_data.csv", index=False)

!pip install pyarrow

df.to_parquet("cleaned_data.parquet", index=False)

import os

csv_size = os.path.getsize("cleaned_data.csv")
parquet_size = os.path.getsize("cleaned_data.parquet")

print("CSV Size:", csv_size, "bytes")
print("Parquet Size:", parquet_size, "bytes")
```
