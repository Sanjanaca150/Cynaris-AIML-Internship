```python
import pandas as pd
import matplotlib.pyplot as plt

# Replace this with your actual Excel file name
df = pd.read_excel("/content/Indian Population Dataset.xlsx")

# First 5 rows
print(df.head())

# Dataset information
print("\nDataset Info:")
df.info()

# Statistical summary
print("\nSummary Statistics:")
print(df.describe())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

df.describe()

df.isnull().sum()

import matplotlib.pyplot as plt

# Distribution plots for numeric columns
numeric_cols = df.select_dtypes(include='number').columns

for col in numeric_cols:
    plt.figure(figsize=(6,4))
    plt.hist(df[col], bins=10)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()

# Correlation Heatmap
corr = df[numeric_cols].corr()

plt.figure(figsize=(5,4))
plt.imshow(corr, cmap='coolwarm')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Correlation Heatmap")
plt.show()

# Top 10 Category Counts
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    plt.figure(figsize=(8,4))
    df[col].value_counts().head(10).plot(kind='bar')
    plt.title(f"Top 10 {col}")
    plt.ylabel("Count")
    plt.show()
```

The dataset contains 10 Indian states, with four columns: State, Region, Population, and Area_km2.

The dataset has two categorical columns (State, Region) and two numerical columns (Population, Area_km2), making it suitable for Exploratory Data Analysis.

Uttar Pradesh has the highest population in the dataset (approximately 200 million), while Delhi has the smallest area (1,484 km²).

The states belong to four different regions: North, South, East, and West, with multiple states represented in each region.

Population and area vary considerably across states, indicating significant differences in demographic size and geographical coverage.
